```python
import numpy as np

# Load the data from the file
data = np.loadtxt('fNLgNLhNL_test.txt', delimiter=' ')

# Assign the data to variables
k_, A, FNL, ob, ng1, total = data.T
```

```python
from plotfunc import Omega,readdata
s = 1.0
readdata(sigma=s)
```

```python
from tqdm import tqdm

for i in tqdm(range(len(k_))):
    # print(k_[i], A[i], FNL[i])
    myOmega = Omega(k_=k_[i],sigma=s,A=A[i],FNL=FNL[i])
    ob_py = myOmega.ob()
    ng1_py = myOmega.ng1()
    total_py = myOmega.total()
    if np.isclose(ob[i], ob_py, rtol=1e-5) and np.isclose(ng1[i], ng1_py, rtol=1e-5) and np.isclose(total[i], total_py, rtol=1e-5):
        continue
    else:
        print(k_[i], A[i], FNL[i])
        print(f"ob and ob_py are NOT similar for index {i}: ob = {ob[i]}, ob_py = {ob_py}")
        print(f"ng1 and ng1_py are NOT similar for index {i}: ng1 = {ng1[i]}, ng1_py = {ng1_py}")
        print(f"total and total_py are NOT similar for index {i}: total = {total[i]}, total_py = {total_py}")
        break
```
