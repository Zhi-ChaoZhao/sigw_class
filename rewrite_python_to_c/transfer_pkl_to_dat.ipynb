```python
import pickle
import numpy as np

# load hNL-SIGW-1.pkl
with open('hNL-SIGW-1.pkl', 'rb') as f:
    data = pickle.load(f)

import scipy.interpolate as interp
```

```python
data.keys()
```

```python
for function_name in data.keys():
    function = data[function_name]
    knots = function.get_knots()
    x_min = knots[0]
    x_max = knots[-1]
    x = np.linspace(x_min, x_max, 10000)
    y = function(x)
    np.savetxt("./dat/"+function_name + '.dat', np.column_stack((x, y)))
```
