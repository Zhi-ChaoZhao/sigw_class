```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
cl_file = './output/cgwb00_cl.dat'
cls_ad_file = './output/cgwb00_cls_ad.dat'
cls_gwi_file = './output/cgwb00_cls_gwi.dat'
OmegaGW_file = './output/cgwb00_OmegaGW.dat'

cl_data = np.loadtxt(cl_file,comments='#')
cl_l, cl_tt, cl_g1g1, cl_tg1 = cl_data[:,0], cl_data[:,1], cl_data[:,2], cl_data[:,3]

cls_ad_data = np.loadtxt(cls_ad_file,comments='#')
cls_ad_l, cls_ad_tt, cls_ad_g1g1, cls_ad_tg1 = cls_ad_data[:,0], cls_ad_data[:,1], cls_ad_data[:,2], cls_ad_data[:,3]

cls_gwi_data = np.loadtxt(cls_gwi_file,comments='#')
cls_gwi_l, cls_gwi_tt, cls_gwi_g1g1, cls_gwi_tg1 = cls_gwi_data[:,0], cls_gwi_data[:,1], cls_gwi_data[:,2], cls_gwi_data[:,3]

OmegaGW_data = np.loadtxt(OmegaGW_file,comments='#')
OmegaGW_f, OmegaGW_OmegaGW = OmegaGW_data[:,0], OmegaGW_data[:,1]
```

```python
# plot OmegaGW
plt.figure(figsize=(10, 6))
plt.loglog(OmegaGW_f, OmegaGW_OmegaGW*0.6736**2, label='OmegaGW')
plt.xlabel('Frequency [Hz]')
plt.ylabel('OmegaGW')
plt.title('OmegaGW')
plt.legend()
plt.grid(True)
plt.show()
```
