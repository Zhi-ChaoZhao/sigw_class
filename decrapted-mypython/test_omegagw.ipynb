```python
from plotfunc import Omega,readdata,h
import numpy as np
from scipy.interpolate import interp1d
```

```python
f_ggf, d_ggf = np.loadtxt('ggf.dat', unpack=True)
B = interp1d(f_ggf, d_ggf, kind='linear')
```

```python
s = 1.0
A_star = 0.0046
f_star = 0.013
f_NL = 0
f = np.logspace(-5,1,1000)
k_ = f/f_star
readdata(sigma=s)
```

```python
MyOmega = Omega(k_=k_,sigma=s,A=A_star,FNL=f_NL*3/5)
```

```python
results_py = MyOmega.ob()/h**2 * B(f)
```

```python
results_c_f, results_c_f_v= np.loadtxt('/workspace/Ubuntu/SGWBxCMB/class_public/output/cgwb00_OmegaGW.dat',unpack=True)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.loglog(f,results_py,label='python',linestyle='--')
plt.loglog(results_c_f,results_c_f_v,label='c')
plt.legend()
plt.show()
```
