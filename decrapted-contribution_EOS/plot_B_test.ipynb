```python
import numpy as np
import matplotlib.pyplot as plt

# Load data from ggf.dat
f_ggf, d_ggf = np.loadtxt('ggf.dat', unpack=True)

# Load data from test_interp_clang.dat
f_test, d_test = np.loadtxt('test_interp_clang.dat', unpack=True)

# Plot the data
plt.figure(figsize=(10, 6))
plt.loglog(f_ggf, d_ggf, label='ggf.dat')
plt.loglog(f_test, d_test, label='test_interp_clang.dat',linestyle='--')    
plt.xlabel('Frequency')
plt.ylabel('Amplitude')
plt.title('Comparison of ggf.dat and test_interp_clang.dat')
plt.legend()
plt.grid(True)
plt.show()
```
